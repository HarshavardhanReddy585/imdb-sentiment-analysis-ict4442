# Experiment: IMDb sentiment interim results

This notebook is the reproducible visual evidence for Phase 2. It reads only the verified JSON metrics committed in `results/`, generates confusion-matrix images and a test-score comparison, and does not silently retrain a model.

Status: MLP and CNN completed on the full 20,000/5,000/25,000 protocol; BiLSTM is Harshavardhan's in-progress smoke experiment.

```python
from pathlib import Path
import json
from IPython.display import Image, display

ROOT = Path.cwd()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
RESULTS = ROOT / 'results'
FIGURES = ROOT / 'report' / 'figures'
models = ['mlp', 'cnn', 'bilstm']
metrics = {name: json.loads((RESULTS / f'{name}_metrics.json').read_text()) for name in models}
metrics
```

## Verified metric summary

The table below is generated from the same JSON files used by the report. The dataset sizes are shown so that the BiLSTM smoke evidence is not mistaken for a full-protocol result.

```python
for name, result in metrics.items():
    split = result['dataset']
    test = result['test']
    print(f"{result['model']}: train={split['train']}, validation={split['validation']}, test={split['test']} | test accuracy={test['accuracy']:.3f}, F1={test['f1']:.3f}")
```

## Result figures

The PNG figures are generated by `scripts/generate_result_figures.py` from the saved confusion matrices and test metrics. Re-run that script after a new verified training result is written.

```python
for filename in ['verified_test_comparison.png', 'mlp_confusion_matrices.png', 'cnn_confusion_matrices.png', 'bilstm_confusion_matrices.png']:
    print(filename)
    display(Image(filename=str(FIGURES / filename)))
```
